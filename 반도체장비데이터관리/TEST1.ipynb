# 파라미터 바인딩 연습문제

```python
#1
import pymysql
import pandas as pd

conn = pymysql.connect(
    host = 'localhost',
    user = 'root',
    password = 'root',
    database = 'ProcessSensorDB',
    charset='utf8mb4'
)

sensor_name = input("조회할 sensor_name을 입력하세요: ")
query = """
select SE.measurement_id, SE.run_id, S.sensor_name, SE.measured_value, S.unit
from sensormeasurement as SE
join sensor as S on SE.sensor_id = S.sensor_id
where S.sensor_name = %s;
"""
df = pd.read_sql(query, conn, params=(sensor_name,))
print(df)
```

```python
#2
equipment_id = input("조회할 equipment_id를 입력하세요: ")
query = """
select R.run_id, E.equipment_id, E.model_name, P.step_name, R.start_time, R.end_time, R.run_status
from RunHistory as R
join Equipment as E on R.equipment_id = E.equipment_id
join ProcessStep as P on R.step_id = P.step_id
where E.equipment_id = %s
order by start_time;
"""

df = pd.read_sql(query, conn, params=(equipment_id,))
print(df)
```

```python
#3
equipment_id = input("조회할 lot_id를 입력하세요: ")
query = """
select L.lot_id, L.product_name, RH.run_id, P.step_name, P.process_group, E.equipment_id, RH.start_time, RH.run_status
from Lot as L
join RunHistory as RH on L.lot_id = RH.lot_id
join ProcessStep as P on RH.step_id = P.step_id
join Equipment as E on RH.equipment_id = E.equipment_id
where L.lot_id = %s
order by start_time;
"""

df = pd.read_sql(query, conn, params=(equipment_id,))
print(df)
```
