```python
import pymysql
import pandas as pd

#6-1
conn = pymysql.connect(
    host = 'localhost',
    user = 'root',
    password = 'root',
    database = 'ProcessSensorDB',
    charset = 'utf8mb4'
)
equipment_id = input("equipment_id를 입력하세요: ")
query = """
select a.run_id, s.equipment_id, p.step_name, p.process_group, s.sensor_name, sm.measured_value, a.alarm_level
from alarmlog as a 
join sensormeasurement as sm on a.run_id = sm.run_id
join sensor as s on sm.sensor_id = s.sensor_id
join runhistory as r on s.equipment_id = r.equipment_id
join processstep as p on r.step_id = p.step_id
where s.equipment_id = %s;

"""
df = pd.read_sql(query, conn, params=(equipment_id,))
print(df)

# 6-2
df = pd.read_sql(query, conn)
pivot_df = df.pivot_table(
    index = 'equipment_id',
    columns = '',
    values = '',
    aggfunc= ''
)
pivot_df.to_csv(
    "EQ장비번호_operation_report.csv",
    encoding='utf-8-sig'
)

```
