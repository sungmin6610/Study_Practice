```python
number = int(input("숫자를 입력합니다: ")) # 동적 타이핑 변수

if number > 0:
    #print(number, "는 양수입니다.")
    print(f"{number} "는 양수입니다.")
elif number < 0:
    print(number, "는 음수입니다.")
else:
    print("입력한 숫자는 0입니다")
```

```python
number = int(input("숫자를 입력합니다: "))
sum = 0
for  i in range(1, number + 1):
    sum += i
print(f"1부터 {number}까지의 합은 {sum}입니다.)")
```
