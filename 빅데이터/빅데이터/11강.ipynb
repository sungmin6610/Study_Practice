```python
#11-1
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

df = pd.read_csv('D:/2604340036 - 송성민/빅데이터/빅데이터/data/ind_ttest.csv')

df.head()
df.groupby('group').count()
df.groupby('group').mean()
df.groupby('group').boxplot(grid=False)
plt.show()

group_1 = df.loc[df.group=='A', 'height']
group_2 = df.loc[df.group=='B', 'height']
print(group_1)
print(group_2)

stats.shapiro(group_1) #정규성 검정
stats.shapiro(group_2)

stats.levene(group_1, group_2) #등분산성 검정

result = stats.ttest_ind(group_1, group_2, equal_var=True) #독립표본 t-검정
print(result)
```

```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

df = pd.read_csv('D:/2604340036 - 송성민/빅데이터/빅데이터/data/paired_ttest.csv')

df.head()
df[['before', 'after']].mean()
(df['after']-df['before']).mean()

fig, axes = plt.subplots(nrows=1, ncols=2)
df['before'].plot.box(grid=False, ax=axes[0])
plt.ylim([60, 100])
df['after'].plot.box(grid=False, ax=axes[1])
plt.show()

stats.shapiro(df['after']-df['before'])

result = stats.ttest_rel(df['before'], df['after'])
print(result)
```

```python
# 11-4

import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
df = pd.read_csv('D:/2604340036 - 송성민/빅데이터/빅데이터/data/wilcoxon_test.csv')
print(df)

df.mean()
(df['post']-df['pre']).mean()

fig, axes = plt.subplots(nrows=1, ncols=2)
df['pre'].plot.box(grid=False, ax=axes[0])
plt.ylim([60, 100])
df['post'].plot.box(grid=False, ax=axes[1])
plt.show()

stats.wilcoxon(df['post'], df['pre'])
```
