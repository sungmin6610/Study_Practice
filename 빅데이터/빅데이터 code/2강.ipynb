```python
import pandas as pd

# 1. Score DataFrame
score = pd.DataFrame([[85, 96, 40, 95],
                      [73, 69, 45, 80],
                      [78, 50, 60, 90]])
print(score)

score.index = ['John', 'Jane', 'Tom']
score.columns = ['KOR', 'ENG', 'MATH', 'SCI']
print(score)

print(score.iloc[2, 1])           # 50 출력
print(score.loc['Tom', 'ENG'])    # 50 출력

# 2. Age Series (정의 후 인덱싱 실행)
age = pd.Series([25, 34, 19, 45, 60])
print(age)

age.index = ['Jhon', 'Jane', 'Tom', 'Micle', 'Tom']
print(age)

print(age.iloc[3])                # 45 출력
print(age.loc['Tom'])             # 중복된 Tom의 값(19, 60) 출력

# 3. Population Series
population = pd.Series([523, 675, 690, 720, 800])
print(population)

population.index = [2021, 2022, 2023, 2024, 2025]
print(population)

print(population.iloc[2])         # 위치 기반(3번째 값) -> 690 출력
print(population.loc[2023])       # 레이블 기반(2023년 값) -> 690 출력
```

```python
import pandas as pd

# 1. 데이터를 'data'라는 변수에 바로 할당 (중간의 1,6, 오타 수정)
data = pd.DataFrame([
    [-0.1, 0.0, -0.1, -0.2], 
    [1.8, 2.0, 1.6, 1.6], 
    [6.2, 6.8, 5.8, 5.9], 
    [12.3, 12.9, 11.5, 11.5], 
    [17.9, 18.5, 17.1, 17.1], 
    [22.2, 22.8, 21.6, 21.5]
])

# 2. 인덱스와 컬럼 설정
data.index = ['1월', '2월', '3월', '4월', '5월', '6월']  
data.columns = ['전북', '전주', '군산', '부안']

# 3. 데이터 확인 (출력하려면 print를 사용하세요)
print(data.iloc[2, 1])      # 3월 전주 값 (6.8)
print(data.loc['3월', '전주']) # 위와 동일한 결과
print(data.iloc[3, 3])      # 4월 부안 값 (11.5)
print(data.loc['4월', '부안']) # 위와 동일한 결과
                      
```
