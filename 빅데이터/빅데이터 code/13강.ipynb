```python
from SQLite import *

SL = SQLite()

sql = 'select 호선명, 역번호, 승객유형, "07-08시간대" from subway limit 10'
result = SL.run_sql(sql)
result.head()

df_fivot = result.pivot_table(index='호선명', columns='승객유형', values="07-08시간대", aggfunc='sum')

print(df_fivot)

df_fivot.to_csv('subway_summary.csv', encoding='EUC-KR')

SL.close_db()
```
