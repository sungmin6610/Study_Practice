## 1.6.3. try예외 처리(try-except) 외부 시스템과의 통신

```python
# 좋은 예 - 네트워크 오류는 예측할 수 없는 예외 상황
import requests

def fetch_user_data(user_id):
    try:
        response = requests.get(f"https://api.example.com/users/{user_id}")
        return response.json()
    except requests.RequestException as e:
        print(f"네트워크 오류가 발생했습니다: {e}")
        return None
    
user_data = fetch_user_data(123)
if user_data:
    print(user_data)
else:
    print("데이터를 가져오지 못했습니다.")
```

## 파일/데이터베이스 접근

```python
# 좋은 예 - 파일 읽기 중 발생할 수 있는 다양한 예외들
def read_file(filename):
    try:
        with open(filename, 'r') as f:
            return f.read()
    except FileNotFoundError:
        print(f"파일을 찾을 수 없습니다: {filename}")
    except PermissionError:
        print(f"파일에 접근할 권한이 없습니다: {filename}")
    except Exception as e:
        print(f"파일 읽기 중 오류가 발생했습니다: {e}")
    return None

content = read_file("data.txt")
if content:
    print(content)
else:
    print("파일을 읽어오지 못했습니다.")
```

## 데이터 변환/파싱

```python
# 좋은 예 - 문자열을 숫자로 변환할 때의 예외
def safe_int_conversion(value):
    try:
        return int(value)
    except ValueError:
        print(f"'{value}'는 숫자로 변환할 수 없습니다.")
        return None
```

## 1.6.4. 잘못된 사용 예시와 개선

```python
# 나쁜 예 - 키워드 없음은 예상 가능한 상황인데 예외로 처리
def search_books(keyword):
    try:
        if keyword.strip() == "":
            raise ValueError("키워드가 비어있습니다")
        return perform_search(keyword)
    except ValueError as e:
        print(f"오류: {e}")
        return []
```

```python
# 좋은 예 - 예상 가능한 상황은 조건문으로
def search_books(keyword):
    if keyword is None or keyword.strip() == "":
        print("오류: 검색 키워드가 필요합니다.")
        return []

    return perform_search(keyword)
```

```python
# 나쁜 예 - 모든 가능한 예외를 미리 체크하려고 함
def divide_numbers(a, b):
    if isinstance(a, str) or isinstance(b, str):
        return "오류: 문자열은 나눌 수 없습니다"
    if b == 0:
        return "오류: 0으로 나눌 수 없습니다"
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        return "오류: 숫자가 아닙니다"

    return a / b
```

```python
# 좋은 예 - 예외가 발생할 수 있는 연산은 try-except로
def divide_numbers(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "오류: 0으로 나눌 수 없습니다"
    except TypeError:
        return "오류: 숫자만 나눌 수 있습니다"
```
