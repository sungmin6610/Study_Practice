```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
import joblib  # 모델 저장용

# 1. 데이터 로드
train_df = pd.read_csv('D:/2604340036 - 송성민/머신러닝/머신러닝 시험대비/train.csv')

# 2. [조건 1] 5개 Feature 선정 및 조합 규칙 정의
# 예시: 중요도가 높은 3개 변수 선택 + 2개의 변수를 조합하여 새 변수 생성
def preprocess_features(df):
    new_df = pd.DataFrame()
    
    # 기존 Feature 중 3개 선택 (예시 피처명: feat_1, feat_2, feat_3)
    new_df['feat_1'] = df['feat_1']
    new_df['feat_2'] = df['feat_2']
    new_df['feat_3'] = df['feat_3']
    
    # 새롭게 조합한 Feature 2개 생성
    new_df['feat_comb_1'] = df['feat_4'] + df['feat_5']
    new_df['feat_comb_2'] = df['feat_6'] * df['feat_7']
    
    return new_df

# 전처리 적용 (정확히 5개의 Feature만 남김)
X_train_full = preprocess_features(train_df)
y_train_full = train_df['target'] # 타겟 변수명에 맞게 수정

# (선택) 자체 검증을 위한 train/val 분할
X_train, X_val, y_train, y_val = train_test_split(X_train_full, y_train_full, test_size=0.2, random_state=42, stratify=y_train_full)

# 3. [조건 2] 선정된 5개 Feature로만 모델 학습
# macro f1을 높이기 위해 class_weight 조정 고려
model = RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42)
model.fit(X_train, y_train)

# 내부 검증 점수 확인 (macro avg f1-score 확인용)
preds = model.predict(X_val)
print(classification_report(y_val, preds))

# 전체 데이터로 최종 재학습 (성능 극대화)
model.fit(X_train_full, y_train_full)

# 4. [조건 3] 최적 모델 1개 지정하여 파일로 저장
joblib.dump(model, 'best_model.pkl')
print("모델 저장 완료: best_model.pkl")
```
